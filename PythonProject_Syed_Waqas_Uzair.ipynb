**Initializing students' grade dictionary**

```python
student_grades = {}
```

**Defining all the functions**

```python
def add_grade():
  '''Add a new student and their grade to the dictionary.'''

  # student name input
  first_name = input("\nEnter First Name of the Student: ")
  second_name = input("Enter Second Name of the Student: ")
  full_name = first_name + " " + second_name

  # student grade input
  grade = input("Enter Student's grade: ")

  # confirmation message
  student_grades[full_name] = grade
  print(f"The data has been recorded successfully: {full_name} --> {student_grades[full_name]}\n")

#-------------------------------------------------------------------------------

def show_all_grades():
  '''Display all students and their grades.'''

  # message if the dictionary is empty
  if len(student_grades) == 0:
    print("\nGrade list empty. Please add grades.\n")


  # printing all grades
  else:
    print("\nBelow are the grades list:")
    for i in student_grades:
       print(f"{i} : {student_grades[i]}")
    print("\n")

#-------------------------------------------------------------------------------

def show_specific_grade():
  '''Show the grade for a specific student.'''

  # message if the dictionary is empty
  if len(student_grades) == 0:
    print("\nGrade list empty. Please add grades.\n")

  else:

   print("\nExtracting grade of a specific student:\n")

   # asking for the name of the student
   try:
    first_name = input("\nPlease enter the first name of the student: ")
    second_name = input("Please enter the second name of the student: ")
    full_name = first_name + " " + second_name
    print(f"The grade of {full_name} is {student_grades[full_name]}.\n")

   except (KeyError, TypeError):
    print("\nStudent not found.")

   finally:
    print("Program Execution Completed.\n")

#-------------------------------------------------------------------------------

def update_grade():
  '''Update the grade of an existing student.'''

  # asking for the name of the student
  first_name = input("\nPlease enter the first name of the student: ")
  second_name = input("Please enter the second name of the student: ")
  full_name = first_name + " " + second_name

  # checking whether the provided name exists in the dictionary
  if full_name in student_grades:
     new_grade = input(f"Please enter new grade for {full_name}: ")
     student_grades[full_name] = new_grade
     print(f"{full_name} grade successfully updated to {student_grades[full_name]}.\n")

  else:
    print("Student not found.\n")

#-------------------------------------------------------------------------------

def delete_grade():
  '''Delete a student's grade record.'''

    # asking for the name of the student
  first_name = input("\nPlease enter the first name of the student: ")
  second_name = input("Please enter the second name of the student: ")
  full_name = first_name + " " + second_name

  # checking whether the provided name exists in the dictionary
  if full_name in student_grades:
     student_grades.pop(full_name)
     print(f"{full_name} grade successfully deleted.\n")

  else:
    print("Student not found.\n")

#-------------------------------------------------------------------------------

def menu():
  '''Provide a continuous menu for the user.'''

  # defining variable for while loop, initalizing it as True
  i = True

  while(i):

    # generating menu
    print("===== Grade Management System ===== \n\n")
    print("1. Add Student Grade\n")
    print("2. Show All Grades\n")
    print("3. Show Specific Student's Grade\n")
    print("4. Update Grade\n")
    print("5. Delete Grade\n")
    print("6. Exit\n\n")

    # use exceptional handling to make sure user chooses correct option
    try:

     user_selection = int(input("Choose an option (1–6): "))

     if user_selection == 6:
       i = False
       print("\nExiting program. Goodbye!")

     elif user_selection == 1:
       add_grade()

     elif user_selection == 2:
       show_all_grades()

     elif user_selection == 3:
       show_specific_grade()

     elif user_selection == 4:
       update_grade()

     elif user_selection == 5:
       delete_grade()

    except (TypeError, ValueError):
      print("\nPlease enter the correct option i.e from (1-6).\n")

#-------------------------------------------------------------------------------
```

**Program Menu**

```python
menu()
```
