## **Nulls** [8 marks]

[1+1+3+2+1]

Find out how many nulls in each column

```python
# check null count in each column
sales.isnull().sum()
```

Find all those rows where brand is null

```python
# rows where brand is null
sales[sales['brand'].isnull()]
```

Fill all the nulls in brand by extracting the brand name from product_name

(hint: look into str.rsplit)

```python
# fill null brand values by extracting from product_name
sales['brand'] = sales['brand'].fillna(sales['product_name'].str.rsplit(' ', n=1).str[0])
```

```python
int(sales['brand'].isnull().sum())
```

Fill all the nulls in payment_method with the most frequently occuring payment method of the WHOLE column

```python
# fill nulls in payment_method with the most frequent value
sales['payment_method'] = sales['payment_method'].fillna(sales['payment_method'].mode()[0])
```

Check total nulls specifically in the payment_method column to see if all were successfully removed

```python
# check nulls in payment_method
int(sales['payment_method'].isnull().sum())
```

### **Total Sales** [3 marks]

Create a total_sale column (remember to factor in discount). It should be rounded to 2 decimal places

```python
# create total_sale column factoring in discount
sales['total_sale'] = round(sales['price'] * sales['quantity'] * (1 - sales['discount_percent'] / 100), 2)
```

### **Groupby and Transform** [6 marks]

[3 + 3]

Use groupby to find out the number of **unique** brands within each subcategory.

Store this grouped data in a variable called brands_per_subcategory.

```python
# find number of unique brands within each subcategory
brands_per_subcategory = sales.groupby('subcategory')['brand'].nunique()
```

```python
# display the grouped data
brands_per_subcategory
```

```python
# display the actual brand names within each subcategory
sales.groupby('subcategory')['brand'].unique()
```

For each null in the price column, replace it with the **median** price of the subcategory occuring at that row.

(hint: use transform to get a column of means that is mappable to our dataframe and use that with fillna)


```python
# fill null prices with the median price of that row's subcategory
subcategory_median_price = sales.groupby('subcategory')['price'].transform('median')
```

```python
sales['price'] = sales['price'].fillna(subcategory_median_price)
```

```python
int(sales['price'].isnull().sum())
```

## **Plotting** [7 marks]

[2+3+2]

Plot a count plot of brands.

**A little detail about the below code block:**
- `figsize=(12, 6)` gives it enough width so the brand names don't get squished together
- `order=sales['brand'].value_counts().index` sorts the bars from most to least frequent brand, instead of the default alphabetical/random order. It makes it easier to read
- `plt.xticks(rotation=45)` angles the brand labels so they don't overlap each other

```python
# count plot of brands
plt.figure(figsize=(12, 6))
sns.countplot(data=sales, x='brand', order=sales['brand'].value_counts().index)
plt.xticks(rotation=45)
plt.show()
```

**Plotting Again But Now with the bar count labels**

-`plt.figure(figsize=(12, 6))` : creates a blank canvas sized 12x6 inches so the brand labels have room and don't overlap.

-`ax = sns.countplot(data=sales, x='brand', order=sales['brand'].value_counts().index)` : plots one bar per brand showing row counts, ordered from most to least frequent, and stores the plot object in ax.

-`ax.bar_label(ax.containers[0])` : writes each bar's count value as text just above that bar.

-`plt.xticks(rotation=45)` : tilts the x-axis brand labels by 45 degrees so they don't overlap each other.

-`plt.show()` : renders and displays the final chart.

```python
plt.figure(figsize=(12, 6))
ax = sns.countplot(data=sales, x='brand', order=sales['brand'].value_counts().index)
ax.bar_label(ax.containers[0])
plt.xticks(rotation=45)
plt.show()
```

Plot a histogram of the prices in our data. Bins should be 0-50, 50-100, 100-150, 150-200, 200-250, 250-300
